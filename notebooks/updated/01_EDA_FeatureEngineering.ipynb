# 01. Exploratory analysis and feature engineering

Rossmann runs over a thousand drugstores in Germany. This notebook looks at the raw Kaggle data
(daily sales per store, 2013 to mid 2015) and then builds the features the forecasting models use.

What to look for: weekly and yearly rhythm, what promotions do to sales, how stores differ, and
the data quirks (closures, a gap in late 2014) that the later notebooks have to handle.

```python
import warnings

warnings.filterwarnings("ignore")

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from rossmann_forecast.config import DEFAULT_ORIGINS, HORIZON, MIN_LAG
from rossmann_forecast.data import load_panel
from rossmann_forecast.metrics import all_metrics, wape
from rossmann_forecast.plotting import COLORS, NAMES, apply_style, find_root
from rossmann_forecast.splits import rolling_origin_folds

apply_style()
ROOT = find_root()
panel = load_panel(ROOT / "data")
folds = rolling_origin_folds(DEFAULT_ORIGINS, HORIZON, panel.dates[-1])
final_fold = folds[-1]
print(f"{len(panel.stores):,} stores, {panel.dates[0].date()} to {panel.dates[-1].date()}")
```

## 1. Raw data

```python
train = pd.read_csv(ROOT / "data" / "train.csv", dtype={"StateHoliday": str}, parse_dates=["Date"])
store = pd.read_csv(ROOT / "data" / "store.csv")
df = train.merge(store, on="Store", how="left")
print(f"train {train.shape}, store {store.shape}, merged {df.shape}")
df.head()
```

```python
closed = (df["Open"] == 0).sum()
zero = ((df["Open"] == 1) & (df["Sales"] == 0)).sum()
print(f"rows with the store closed: {closed:,} ({closed / len(df):.1%})")
print(f"rows open but with zero sales: {zero:,}")
open_df = df[(df["Open"] == 1) & (df["Sales"] > 0)].copy()
print(f"rows kept for modelling and scoring: {len(open_df):,}")
missing = store.isna().sum()
missing[missing > 0].rename("missing values in store.csv").to_frame()
```

Closed days and zero-sales days are not scored, which matches the Kaggle metric. Missing store
attributes are mostly competition and Promo2 fields that simply do not apply to some stores.

## 2. Sales over time

```python
daily = open_df.groupby("Date")["Sales"].mean()
fig, ax = plt.subplots(figsize=(12, 3.8))
ax.plot(daily.index, daily.values, color="#bbb", lw=0.8, label="daily mean per open store")
ax.plot(daily.index, daily.rolling(28, center=True).mean(), color=COLORS["lightgbm"], lw=2, label="28-day average")
ax.set_ylabel("Sales per open store")
ax.set_title("Average sales per open store over time")
ax.legend()
fig.tight_layout()
plt.show()
```

Sales sit around 6,000 to 7,500 per store-day with a sharp December peak each year and a visible
weekly saw-tooth. Both patterns are things a forecast has to learn.

## 3. Weekly rhythm and promotions

```python
names = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
data = [open_df.loc[open_df["DayOfWeek"] == d, "Sales"] for d in range(1, 8)]
axes[0].boxplot(data, tick_labels=names, showfliers=False)
axes[0].set_title("Sales by day of week (open days)")
axes[0].set_ylabel("Sales")
g = open_df.groupby(["DayOfWeek", "Promo"])["Sales"].mean().unstack()
x = np.arange(7)
axes[1].bar(x - 0.2, g[0].reindex(range(1, 8)).values, 0.4, label="no promo", color="#999")
axes[1].bar(x + 0.2, g[1].reindex(range(1, 8)).values, 0.4, label="promo", color=COLORS["lightgbm"])
axes[1].set_xticks(x, names)
axes[1].set_title("Average sales with and without a promotion")
axes[1].legend()
fig.tight_layout()
plt.show()
promo_share = open_df.groupby("DayOfWeek")["Promo"].mean().round(2).rename("share of open days with a promo")
promo_share.index = names
promo_share.to_frame().T
```

Sundays (mostly closed, and when open, busy) and Mondays are the big days. Promos only run
Monday to Friday and lift every weekday. Notebook 05 looks at what that lift means and how far it
can be trusted.

## 4. Seasonality

```python
open_df["Year"] = open_df["Date"].dt.year
open_df["Month"] = open_df["Date"].dt.month
m = open_df.groupby(["Year", "Month"])["Sales"].mean().unstack(0)
fig, ax = plt.subplots(figsize=(10, 3.8))
for year in m.columns:
    ax.plot(m.index, m[year], marker="o", label=str(year))
ax.set_xticks(range(1, 13))
ax.set_xlabel("Month")
ax.set_ylabel("Average sales per open store")
ax.set_title("Monthly profile by year")
ax.legend()
fig.tight_layout()
plt.show()
```

## 5. Store characteristics

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 3.8))
for ax, col in zip(axes[:2], ["StoreType", "Assortment"], strict=True):
    s = open_df.groupby(col)["Sales"].mean()
    ax.bar(s.index, s.values, color=COLORS["lightgbm"])
    ax.set_title(f"Average sales by {col}")
sh = open_df.groupby("StateHoliday")["Sales"].mean().rename({"0": "none", "a": "public", "b": "Easter", "c": "Christmas"})
axes[2].bar(sh.index, sh.values, color="#8e6bbf")
axes[2].set_title("Average sales by state holiday (open stores)")
fig.tight_layout()
plt.show()
```

```python
by_store = open_df.groupby("Store").agg(sales=("Sales", "mean"), dist=("CompetitionDistance", "first"), promo2=("Promo2", "first"))
fig, axes = plt.subplots(1, 2, figsize=(13, 3.8))
axes[0].scatter(by_store["dist"], by_store["sales"], s=6, alpha=0.5, color=COLORS["lightgbm"])
axes[0].set_xscale("log")
axes[0].set_xlabel("Distance to nearest competitor (m, log scale)")
axes[0].set_ylabel("Store average daily sales")
axes[0].set_title("Competition distance versus store size")
p2 = by_store.groupby("promo2")["sales"].mean()
axes[1].bar(["no Promo2", "Promo2"], p2.values, color=["#999", COLORS["lightgbm"]])
axes[1].set_title("Stores in the continuing Promo2 programme")
fig.tight_layout()
plt.show()
```

## 6. Data quirks the models must handle

```python
share_open = panel.open.mean(axis=1)
rows_per_day = train.groupby("Date")["Store"].nunique()
fig, axes = plt.subplots(2, 1, figsize=(12, 5), sharex=True)
axes[0].plot(share_open.index, share_open.values, color="#555", lw=0.8)
axes[0].set_ylabel("Share of stores open")
axes[0].set_title("Sundays and holidays close most stores")
axes[1].plot(rows_per_day.index, rows_per_day.values, color=COLORS["lightgbm"])
axes[1].set_ylabel("Stores with a record")
axes[1].set_title("About 180 stores have no records between July and December 2014")
fig.tight_layout()
plt.show()
```

The late-2014 gap (stores closed for refurbishment) means the first forecast window has fewer
stores than the others. The pipeline scores only days a store was open with positive sales.

## 7. Feature engineering

Every sales-based feature is **lagged by at least the 42-day forecast horizon**. A feature for
date `t` may only read sales up to `t - 42`, so it is available to anyone forecasting `t` six weeks
ahead. Calendar, promo, holiday and open/closed information for `t` is planned in advance and is
used as is.

```python
from rossmann_forecast.features import build_features, feature_columns

feats = build_features(panel, MIN_LAG)
cols = feature_columns(feats)
groups = {
    "sales history, at least 42 days old": [c for c in cols if c.startswith(("roll", "dow_", "ly_", "promo_ratio"))],
    "promo and holiday calendar (known in advance)": [c for c in cols if "promo" in c and not c.startswith(("roll", "promo_ratio"))] + [c for c in cols if "holiday" in c],
    "calendar": ["dow", "day", "month", "week", "doy"],
    "store attributes": ["store", "store_type", "assortment", "comp_distance", "comp_open_months", "promo2_active"],
    "open/closed neighbours": ["open_prev", "open_next"],
}
pd.DataFrame({"features": {k: ", ".join(v) for k, v in groups.items()}})
```

```python
one = feats[feats["store_id"] == 1].set_index("date")
window = one.loc["2014-09-01":"2015-03-31"]
fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(window.index, window["sales"], color="#bbb", lw=0.8, label="actual sales")
ax.plot(window.index, window["roll28_mean"], color="#1f77b4", label="28-day mean, 42+ days old")
ax.plot(window.index, window["roll91_mean"], color=COLORS["lightgbm"], label="91-day mean, 42+ days old")
ax.plot(window.index, window["ly_mean3"], color=COLORS["last_year"], ls=":", label="last year, same weekday")
ax.set_title("Store 1: sales history features follow the level but never look at the last 6 weeks")
ax.legend(ncol=2, fontsize=8)
fig.tight_layout()
plt.show()
```

```python
sample = feats[(feats["is_open"] == 1) & feats["sales"].notna()].sample(60000, random_state=0)
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, col in zip(axes, ["dow_mean8", "ly_364", "roll91_mean"], strict=True):
    ax.hexbin(sample[col], sample["sales"], gridsize=45, bins="log", cmap="Reds", mincnt=1)
    r = sample[[col, "sales"]].corr().iloc[0, 1]
    ax.set_title(f"{col} (correlation {r:.2f})")
    ax.set_xlabel(col)
axes[0].set_ylabel("actual sales")
fig.tight_layout()
plt.show()
```

## 8. Proof that no future sales leak into the features

```python
import dataclasses

target = pd.Timestamp("2015-03-10")
cutoff = target - pd.Timedelta(days=MIN_LAG)
scrambled_sales = panel.sales.copy()
later = scrambled_sales.index > cutoff
scrambled_sales.loc[later] = scrambled_sales.loc[later] * 7 + 1234
scrambled = dataclasses.replace(panel, sales=scrambled_sales)
feats_scrambled = build_features(scrambled, MIN_LAG)

def on(frame):
    return frame[frame["date"] == target].sort_values("store_id")[cols].reset_index(drop=True)

same = on(feats).equals(on(feats_scrambled))
print(f"Scrambled every sale after {cutoff.date()} (42 days before {target.date()}).")
print(f"Features for {target.date()} identical: {same}")

# positive control: scrambling usable history must change the features
history_scrambled = panel.sales.copy()
usable = (history_scrambled.index > cutoff - pd.Timedelta(days=60)) & (history_scrambled.index <= cutoff)
history_scrambled.loc[usable] = history_scrambled.loc[usable] * 7 + 1234
changed = build_features(dataclasses.replace(panel, sales=history_scrambled), MIN_LAG)
print(f"Scrambling usable history changes the features: {not on(feats).equals(on(changed))}")
```

The same check runs as an automated test (`tests/test_features.py`) on every change. The next
notebook uses these features and the fold design for baselines and per-store models.
